```python
import numpy as np
import pandas as pd

# Employee dataset using NumPy
data = np.array([
    [101, "dhanu", 25, "IT", 35000],
    [102, "ranika", 28, "HR", 40000],
    [103, "tejas", 30, "IT", 50000],
    [104, "pankaj", 26, "Finance", 45000],
    [105, "sakshi", 35, "IT", 60000],
])
```

```python
df=pd.DataFrame(data,columns=["Employee_ID", "Name", "Age", "Department", "Salary"]
)

df
```

```python
df["Employee_ID"] = df["Employee_ID"].astype(int)
df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(int)
df
```

```python
#employee datasets
print("Employee Dataset:")
print(df)
```

```python
print(" Total Employees:")
print(len(df))


```

```python
print("Average Age")
print(df["Age"].mean())

```

```python
print("Average Salary")
print(df["Salary"].mean())
```

```python
print("Average Salary Department-wise:")
print(df.groupby("Department")["Salary"].mean())

```

```python
print("Employees Department-wise:")
print(df["Department"].value_counts())
```
